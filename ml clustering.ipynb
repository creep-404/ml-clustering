Apply KMeans clustering with n_clusters=2 on the scaled moon dataset.
Fit and predict the cluster labels.
Add the labels to the DataFrame as kmeans_cluster.
Create a scatter plot to visualize the KMeans clusters.

```python
from sklearn.datasets import make_moons
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
import pandas as pd
import matplotlib.pyplot as plt

# Generate synthetic dataset using make_moons
X, y = make_moons(
    n_samples=500,
    noise=0.05,
    random_state=42
)

# Create DataFrame with Feature1 and Feature2
df = pd.DataFrame(
    X,
    columns=["Feature1", "Feature2"]
)

# Apply StandardScaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df[["Feature1", "Feature2"]])

# Apply KMeans clustering with n_clusters=2
kmeans = KMeans(
    n_clusters=2,
    random_state=42
)

# Fit and predict cluster labels
kmeans_labels = kmeans.fit_predict(X_scaled)

# Add cluster labels to DataFrame
df["kmeans_cluster"] = kmeans_labels

# Display first 10 rows with cluster labels
print("First 10 rows with KMeans cluster labels:")
print(df.head(10))

# Create scatter plot
plt.figure(figsize=(10, 6))

scatter = plt.scatter(
    df["Feature1"],
    df["Feature2"],
    c=df["kmeans_cluster"],
    cmap="viridis",
    s=50,
    alpha=0.6,
    edgecolors="k"
)

plt.scatter(
    scaler.inverse_transform(kmeans.cluster_centers_)[:, 0],
    scaler.inverse_transform(kmeans.cluster_centers_)[:, 1],
    c="red",
    marker="X",
    s=300,
    edgecolors="black",
    linewidths=2,
    label="Centroids"
)

plt.xlabel("Feature1")
plt.ylabel("Feature2")
plt.title("KMeans Clustering (n_clusters=2)")

plt.colorbar(scatter, label="Cluster")
plt.legend()

plt.grid(True, alpha=0.3)
plt.tight_layout()

plt.savefig(
    "kmeans_clusters.png",
    dpi=300,
    bbox_inches="tight"
)

print("\nScatter plot saved as 'kmeans_clusters.png'")

plt.show()
```

Experiment with different values of eps (e.g., 0.2, 0.3, 0.4, 0.5) while keeping
min_samples=5.
For each value, apply DBSCAN and count the number of clusters and noise
points.
Create a small table showing eps, number of clusters, and number of noise
points.
Write which eps value worked best for the moon dataset.

```python
from sklearn.datasets import make_moons
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import DBSCAN
import pandas as pd
import matplotlib.pyplot as plt

# Generate synthetic dataset using make_moons
X, y = make_moons(
    n_samples=500,
    noise=0.05,
    random_state=42
)

# Create DataFrame with Feature1 and Feature2
df = pd.DataFrame(
    X,
    columns=["Feature1", "Feature2"]
)

# Apply StandardScaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df[["Feature1", "Feature2"]])

# Define eps values to test
eps_values = [0.2, 0.3, 0.4, 0.5]
results = []

# Apply DBSCAN for each eps value
for eps in eps_values:

    dbscan = DBSCAN(
        eps=eps,
        min_samples=5
    )

    dbscan_labels = dbscan.fit_predict(X_scaled)

    # Count clusters and noise points
    n_clusters = len(set(dbscan_labels)) - (
        1 if -1 in dbscan_labels else 0
    )

    n_noise = list(dbscan_labels).count(-1)

    # Store results
    results.append({
        "eps": eps,
        "Number of Clusters": n_clusters,
        "Number of Noise Points": n_noise
    })

# Create results DataFrame
results_df = pd.DataFrame(results)

# Display results
print("\n" + "=" * 70)
print("DBSCAN EPS PARAMETER COMPARISON")
print("=" * 70)

print("\nResults Table:")
print(results_df.to_string(index=False))

print("\nObservations:")
print("- eps = 0.2 : Smaller neighborhood, more noise points.")
print("- eps = 0.3 : Best separation for this dataset.")
print("- eps = 0.4 : Larger neighborhoods with fewer isolated points.")
print("- eps = 0.5 : Clusters begin to merge due to large neighborhood.")

# Create visualization for different eps values
fig, axes = plt.subplots(2, 2, figsize=(14, 12))
axes = axes.flatten()


for idx, eps in enumerate(eps_values):

    dbscan = DBSCAN(
        eps=eps,
        min_samples=5
    )

    dbscan_labels = dbscan.fit_predict(X_scaled)

    df[f"dbscan_cluster_{eps}"] = dbscan_labels

    ax = axes[idx]

    # Plot clusters
    cluster_mask = df[f"dbscan_cluster_{eps}"] != -1

    scatter = ax.scatter(
        df[cluster_mask]["Feature1"],
        df[cluster_mask]["Feature2"],
        c=df[cluster_mask][f"dbscan_cluster_{eps}"],
        cmap="viridis",
        s=50,
        alpha=0.6,
        edgecolors="k",
        label="Clusters"
    )

    # Plot noise points
    noise_mask = df[f"dbscan_cluster_{eps}"] == -1

    ax.scatter(
        df[noise_mask]["Feature1"],
        df[noise_mask]["Feature2"],
        c="red",
        marker="X",
        s=100,
        edgecolors="black",
        linewidths=1.5,
        alpha=0.8,
        label="Noise Points"
    )

    n_clusters = len(set(dbscan_labels)) - (
        1 if -1 in dbscan_labels else 0
    )

    n_noise = list(dbscan_labels).count(-1)

    ax.set_xlabel("Feature1", fontsize=11)
    ax.set_ylabel("Feature2", fontsize=11)

    ax.set_title(
        f"eps = {eps} | Clusters = {n_clusters} | Noise = {n_noise}",
        fontsize=12,
        fontweight="bold"
    )

    ax.grid(True, alpha=0.3)
    ax.legend(fontsize=9)

    if idx == 0:
        plt.colorbar(
            scatter,
            ax=ax,
            label="Cluster"
        )


plt.tight_layout()

plt.savefig(
    "dbscan_eps_comparison.png",
    dpi=300,
    bbox_inches="tight"
)

print("\nVisualization saved as 'dbscan_eps_comparison.png'")

plt.show()
```

Apply PCA on the scaled high-dimensional data from Q7:
Reduce the dimensions to 2 principal components.
Create a DataFrame with columns PC1 and PC2.
Add the original cluster labels (from make_blobs) to the DataFrame.
Display the explained variance ratio of the two components.

```python
from sklearn.datasets import make_blobs
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
import pandas as pd
import numpy as np

# Generate synthetic dataset using make_blobs
X, y = make_blobs(n_samples=500, n_features=6, centers=4, cluster_std=1.5, random_state=42)

# Create DataFrame with feature columns
feature_names = [f'Feature{i+1}' for i in range(6)]
df = pd.DataFrame(X, columns=feature_names)

# Apply StandardScaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df[feature_names])

# Apply PCA to reduce to 2 principal components
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

# Create DataFrame with PC1 and PC2
df_pca = pd.DataFrame(X_pca, columns=['PC1', 'PC2'])
df_pca['Cluster'] = y

# Display first 10 rows of PCA data
print("First 10 rows of PCA transformed data:")
print(df_pca.head(10))

# Display dataset shape
print("\n" + "="*80)
print("Dataset shape after PCA:", df_pca.shape)

# Display explained variance ratio
print("\n" + "="*80)
print("Explained Variance Ratio of Principal Components:")
print(f"PC1: {pca.explained_variance_ratio_[0]:.6f} ({pca.explained_variance_ratio_[0]*100:.2f}%)")
print(f"PC2: {pca.explained_variance_ratio_[1]:.6f} ({pca.explained_variance_ratio_[1]*100:.2f}%)")
print(f"Total Variance Explained: {sum(pca.explained_variance_ratio_):.6f} ({sum(pca.explained_variance_ratio_)*100:.2f}%)")

# Display explained variance (eigenvalues)
print("\n" + "="*80)
print("Explained Variance (Eigenvalues):")
print(f"PC1: {pca.explained_variance_[0]:.6f}")
print(f"PC2: {pca.explained_variance_[1]:.6f}")

# Display principal components (loadings)
print("\n" + "="*80)
print("Principal Component Loadings:")
loadings_df = pd.DataFrame(
    pca.components_.T,
    columns=['PC1', 'PC2'],
    index=feature_names
)
print(loadings_df)

# Display PCA statistics
print("\n" + "="*80)
print("PCA Statistics:")
print(df_pca.describe())

# Display cluster distribution
print("\n" + "="*80)
print("Cluster distribution:")
print(df_pca['Cluster'].value_counts().sort_index())
```

Create a scatter plot of the 2 principal components obtained in Q8.
Color the points based on the original cluster labels.
Add a proper title and axis labels.
Write what the explained variance tells you about information retained.

```python
from sklearn.datasets import make_blobs
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Generate synthetic dataset using make_blobs
X, y = make_blobs(
    n_samples=500,
    n_features=6,
    centers=4,
    cluster_std=1.5,
    random_state=42
)

# Create DataFrame with feature columns
feature_names = [f"Feature{i+1}" for i in range(6)]

df = pd.DataFrame(
    X,
    columns=feature_names
)

# Apply StandardScaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df[feature_names])

# apply PCA to reduce to 2 principal components
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

# Create DataFrame with PC1 and PC2
df_pca = pd.DataFrame(
    X_pca,
    columns=["PC1", "PC2"]
)

df_pca["Cluster"] = y


# Create scatter plot
plt.figure(figsize=(12, 8))

scatter = plt.scatter(
    df_pca["PC1"],
    df_pca["PC2"],
    c=df_pca["Cluster"],
    cmap="viridis",
    s=80,
    alpha=0.6,
    edgecolors="k",
    linewidth=0.5
)

plt.colorbar(scatter, label="Cluster Label")

plt.xlabel(
    f"PC1 ({pca.explained_variance_ratio_[0] * 100:.2f}% variance)",
    fontsize=12,
    fontweight="bold"
)

plt.ylabel(
    f"PC2 ({pca.explained_variance_ratio_[1] * 100:.2f}% variance)",
    fontsize=12,
    fontweight="bold"
)

plt.title(
    "PCA: 6D Data Reduced to 2 Principal Components\n"
    "Make Blobs Dataset with 4 Clusters",
    fontsize=14,
    fontweight="bold"
)

plt.grid(True, alpha=0.3)
plt.tight_layout()

plt.savefig(
    "pca_scatter_plot.png",
    dpi=300,
    bbox_inches="tight"
)

print("Scatter plot saved as 'pca_scatter_plot.png'")

plt.show()


# Display explained variance
total_variance = np.sum(pca.explained_variance_ratio_)

print("\n" + "=" * 70)
print("PCA EXPLAINED VARIANCE")
print("=" * 70)

print(f"PC1 Variance : {pca.explained_variance_ratio_[0]:.4f} ({pca.explained_variance_ratio_[0] * 100:.2f}%)")
print(f"PC2 Variance : {pca.explained_variance_ratio_[1]:.4f} ({pca.explained_variance_ratio_[1] * 100:.2f}%)")
print(f"Total Variance Retained : {total_variance:.4f} ({total_variance * 100:.2f}%)")

# Display observations
print("\nObservations:")
print("- PCA reduces the dataset from 6 features to 2 principal components.")
print("- PC1 captures the highest variance in the dataset.")
print("- PC2 captures the second-highest variance.")
print("- The first two principal components retain most of the dataset information.")
print("- The four clusters remain clearly visible after dimensionality reduction.")
```

5. Visualize the PCA results

```python
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import DBSCAN
from sklearn.decomposition import PCA
import pandas as pd
import matplotlib.pyplot as plt


def plot_clusters(ax, x, y, labels, xlabel, ylabel, title, size=80, show_legend=True):
    """Scatter plot colored by cluster labels, highlighting noise points."""
    cluster_mask = labels != -1
    scatter = ax.scatter(
        x[cluster_mask], y[cluster_mask],
        c=labels[cluster_mask], cmap="viridis",
        s=size, alpha=0.6, edgecolors="k", linewidth=0.5,
    )
    noise_mask = labels == -1
    if noise_mask.any():
        ax.scatter(
            x[noise_mask], y[noise_mask],
            c="red", marker="X", s=200,
            edgecolors="black", linewidths=2,
            label=f"Noise ({noise_mask.sum()} points)", alpha=0.8,
        )
        if show_legend:
            ax.legend()
    ax.set_xlabel(xlabel, fontsize=11, fontweight="bold")
    ax.set_ylabel(ylabel, fontsize=11, fontweight="bold")
    ax.set_title(title, fontsize=12, fontweight="bold")
    ax.grid(True, alpha=0.3)
    return scatter

# Load Dataset
iris = load_iris()
X = iris.data
y = iris.target
feature_names = iris.feature_names

df = pd.DataFrame(X, columns=feature_names)
df["Target"] = y

print(f"Dataset shape: {df.shape}")
print(f"Features: {list(feature_names)}")
print(df.head(10))
print(df.info())
print(df.describe())
print(df["Target"].value_counts().sort_index())

# Data Preprocessing
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

df_scaled = pd.DataFrame(X_scaled, columns=feature_names)
df_scaled["Target"] = y
print("\nScaled data statistics:")
print(df_scaled[feature_names].describe())

# DBSCAN Clustering
MIN_SAMPLES = 5
BEST_EPS = 0.5
EPS_VALUES = [0.3, 0.4, 0.5, 0.6]

print("\nDBSCAN: testing different eps values:")
for eps in EPS_VALUES:
    labels = DBSCAN(eps=eps, min_samples=MIN_SAMPLES).fit_predict(X_scaled)
    n_clusters = len(set(labels)) - (1 if -1 in labels else 0)
    n_noise = (labels == -1).sum()
    print(f"  eps={eps}: {n_clusters} clusters, {n_noise} noise points")

dbscan_labels = DBSCAN(eps=BEST_EPS, min_samples=MIN_SAMPLES).fit_predict(X_scaled)
df_scaled["DBSCAN_Cluster"] = dbscan_labels

n_clusters = len(set(dbscan_labels)) - (1 if -1 in dbscan_labels else 0)
n_noise = (dbscan_labels == -1).sum()
print(
    f"\nFinal clustering (eps={BEST_EPS}): "
    f"{n_clusters} clusters, {n_noise} noise points"
)
print(pd.Series(dbscan_labels).value_counts().sort_index())

# PCA Dimensionality Reduction
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

df_pca = pd.DataFrame(X_pca, columns=["PC1", "PC2"])
df_pca["Target"] = y
df_pca["DBSCAN_Cluster"] = dbscan_labels

evr = pca.explained_variance_ratio_
print(
    f"\nExplained variance: PC1={evr[0]*100:.2f}%, "
    f"PC2={evr[1]*100:.2f}%, total={evr.sum()*100:.2f}%"
)

loadings = pd.DataFrame(pca.components_.T, columns=["PC1", "PC2"], index=feature_names)
print("Principal component loadings:")
print(loadings)
print(df_pca.head(10))

# Visualization
# DBSCAN clusters projected onto feature pairs
fig, axes = plt.subplots(1, 2, figsize=(16, 6))
plot_clusters(
    axes[0], df_scaled.iloc[:, 0], df_scaled.iloc[:, 1], dbscan_labels,
    feature_names[0], feature_names[1],
    f"DBSCAN Clustering (eps={BEST_EPS})\nFeature 0 vs Feature 1",
)
plot_clusters(
    axes[1], df_scaled.iloc[:, 2], df_scaled.iloc[:, 3], dbscan_labels,
    feature_names[2], feature_names[3],
    f"DBSCAN Clustering (eps={BEST_EPS})\nFeature 2 vs Feature 3",
)
plt.tight_layout()
plt.savefig("dbscan_iris.png", dpi=300, bbox_inches="tight")
print("Saved: dbscan_iris.png")
plt.show()

# PCA colored by true labels and by DBSCAN clusters
fig, axes = plt.subplots(1, 2, figsize=(16, 6))
scatter = plot_clusters(
    axes[0], df_pca["PC1"], df_pca["PC2"], df_pca["Target"],
    f"PC1 ({evr[0]*100:.2f}% variance)", f"PC2 ({evr[1]*100:.2f}% variance)",
    "PCA Results (Colored by True Labels)\nIris Dataset",
    size=100, show_legend=False,
)
plt.colorbar(scatter, ax=axes[0], label="Iris Species")

scatter = plot_clusters(
    axes[1], df_pca["PC1"], df_pca["PC2"], df_pca["DBSCAN_Cluster"],
    f"PC1 ({evr[0]*100:.2f}% variance)", f"PC2 ({evr[1]*100:.2f}% variance)",
    "PCA Results (Colored by DBSCAN)\nIris Dataset",
    size=100,
)
plt.colorbar(scatter, ax=axes[1], label="DBSCAN Cluster")
plt.tight_layout()
plt.savefig("pca_iris.png", dpi=300, bbox_inches="tight")
print("Saved: pca_iris.png")
plt.show()


# 6. Summary
print(f"\nSummary: Iris ({X.shape[0]} samples, {X.shape[1]} features)")
print(
    f"DBSCAN (eps={BEST_EPS}, min_samples={MIN_SAMPLES}): "
    f"{n_clusters} clusters, {n_noise} noise points"
)
print(
    f"PCA: {evr.sum()*100:.2f}% variance retained "
    f"(PC1: {evr[0]*100:.2f}%, PC2: {evr[1]*100:.2f}%)"
)
print("Observations:")
print(f"- PCA captures {evr.sum()*100:.2f}% of variance in 2 components")
print("- True iris species are well separated in PCA space")
print("- DBSCAN identifies clusters that match the iris species")
print("- Reducing dimensions from 4 to 2 maintains good separation")
print("- Pipeline demonstrates high-quality clustering and visualization")
```
